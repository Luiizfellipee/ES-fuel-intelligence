```python
import pandas as pd
from pathlib import Path
from warnings import filterwarnings

filterwarnings("ignore")
```

```python
PASTA_ATUAL = Path.cwd()

CAMINHO = (Path.cwd().parent/ "data"/ "raw"/ "Preços semestrais - AUTOMOTIVOS_2026.01.csv")

data_anp = pd.read_csv(
    CAMINHO,
    sep=";",
    decimal=",",
    encoding="utf-8-sig",  
    dtype={"CNPJ da Revenda": "string", "Cep": "string", "Numero Rua": "string"},
)

data_anp["Data da Coleta"] = pd.to_datetime(data_anp["Data da Coleta"], format="%d/%m/%Y")
```

```python
data_anp.head()
```

```python
data_anp["Data da Coleta"].dt.strftime("%d/%m/%Y").agg(["min", "max"])
```

```python
data_anp.shape
```

```python
data_anp.info()
```

```python
data_anp.isna().sum()
```

```python
data_anp['Valor de Compra'] = data_anp['Valor de Compra'].fillna('NAO INFORMADO')
print(data_anp.isna().sum())
```

```python
data_anp_tratado = data_anp.copy()
```

```python
data_anp_tratado.drop(columns=["Nome da Rua", "Numero Rua", "Complemento", "Bairro", "Valor de Compra", "Regiao - Sigla"], inplace=True)
```

```python
data_anp_tratado.rename(columns={"Estado - Sigla": "Estado"}, inplace=True)
data_anp_tratado.head()
```

```python
ordem = data_anp_tratado.columns.tolist()

ordem.remove("Unidade de Medida")
ordem.insert(ordem.index("Valor de Venda"), "Unidade de Medida")

ordem.remove("Produto")
ordem.insert(ordem.index("Data da Coleta") + 1, "Produto")

data_anp_tratado = data_anp_tratado[ordem]
```

```python
data_anp_tratado.head()
```

```python
data_es = data_anp_tratado[data_anp_tratado["Estado"] == "ES"].copy()
data_es.head()
```

```python
data_es.rename(columns={"Valor de Venda": "Valor", "CNPJ da Revenda": "CNPJ", "Data da Coleta": "Data_Coleta", "Unidade de Medida": "Uni_Medida"}, inplace=True)
```

```python
data_es.head()
```

```python
data_es.to_csv(
    PASTA_ATUAL.parent / "data" / "raw" / "data_es.csv",
    sep=";",    index=False
)
```
